```python
import pandas as pd
import numpy as np
import openpyxl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("Crime_Data_Cleaned.xlsx")
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
# Crime Type Distribution
plt.figure(figsize=(12,5))
sns.countplot(y=df["Primary Type"], order=df["Primary Type"].value_counts().index, palette="viridis")
plt.title("Crime Type Distribution")
plt.show()
```

```python
# Histogram for Crime Frequency by Year
plt.figure(figsize=(8,4))
sns.histplot(df["Year"], bins=20, kde=True, color="blue")
plt.title("Crime Frequency by Year")
plt.show()
```

```python
plt.figure(figsize=(12,5))
sns.barplot(x=df["Primary Type"], y=df["Arrest"].astype(int), estimator=np.mean, palette="coolwarm")
plt.xticks(rotation=90)
plt.title("Arrest Rate by Crime Type")
plt.show()

```

```python
plt.figure(figsize=(12,5))
sns.countplot(y=df["Location Description"], order=df["Location Description"].value_counts().index[:10], palette="magma")
plt.title("Top 10 Crime Locations")
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])  # Convert Date column to datetime format
df["Month_Year"] = df["Date"].dt.to_period("M")  # Extract Year-Month

plt.figure(figsize=(12,5))
df.groupby("Month_Year").size().plot(kind="line", marker="o", color="red")
plt.title("Crime Trends Over Time")
plt.xlabel("Time (Year-Month)")
plt.ylabel("Crime Count")
plt.xticks(rotation=45)
plt.show()
```
